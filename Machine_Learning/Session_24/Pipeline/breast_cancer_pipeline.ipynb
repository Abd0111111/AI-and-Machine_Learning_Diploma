# Breast Cancer Model — باستخدام Pipeline

```python
import pandas as pd
import numpy as np

df = pd.read_csv(r'D:\AI-and-Machine_Learning_Diploma\Machine_Learning\Session_23\breast_cancer_model\breast_cancer.csv.xls')
df
```

```python
from sklearn.preprocessing import LabelEncoder

y = df['diagnosis']
X = df.drop(columns=['id', 'Unnamed: 32', 'diagnosis'], axis=1)

le = LabelEncoder()
y = le.fit_transform(y)

print('X shape:', X.shape)
print('Classes:', le.classes_)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)

print('Train size:', X_train.shape)
print('Test size :', X_test.shape)
```

## بناء الـ Pipeline

بدل ما نحفظ `StandardScaler` و `LogisticRegression` كـ joblib files منفصلين،  
بنلفّهم في **Pipeline واحد** → فايل واحد بس يتحفظ.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('classifier', LogisticRegression())
])

pipeline.fit(X_train, y_train)
pipeline
```

## التقييم

```python
from sklearn.metrics import confusion_matrix, classification_report

y_pred = pipeline.predict(X_test)

cm = confusion_matrix(y_test, y_pred)
cr = classification_report(y_test, y_pred, target_names=le.classes_)

print(cm)
print(cr)
```

## Threshold Analysis

```python
# الـ Pipeline بيعمل الـ scaling تلقائياً قبل predict_proba
y_pred_proba = pipeline.predict_proba(X_test)[:, 1]

# تجربة threshold = 0.4
y_pred_threshold = (y_pred_proba >= 0.4).astype(int)
print(confusion_matrix(y_test, y_pred_threshold))
print(classification_report(y_test, y_pred_threshold, target_names=le.classes_))
```

```python
threshold = np.arange(0.1, 1, 0.05)
results = []

for i in threshold:
    y_pred_t = (y_pred_proba >= i).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test, y_pred_t).ravel()
    results.append({
        'threshold' : i,
        'tn'        : tn,
        'fp'        : fp,
        'fn'        : fn,
        'tp'        : tp,
        'accuracy'  : (tp + tn) / (tp + tn + fp + fn),
        'precision' : tp / (tp + fp),
        'recall'    : tp / (tp + fn),
        'f1_score'  : (fn + tp) / (2 * fn + tp),
    })

results_df = pd.DataFrame(results)
results_df
```

## حفظ الـ Pipeline — فايل واحد بس!

بدل تلت files (`classifier.joblib`, `scaler.joblib`, `label_encoder.joblib`)  
بنحفظ **فايلين بس**: الـ Pipeline + الـ LabelEncoder.

```python
import joblib

joblib.dump(pipeline, 'pipeline.joblib')      # scaler + classifier في واحد
joblib.dump(le,       'label_encoder.joblib') # اللي بيحول B/M لـ 0/1

print('Pipeline saved!')
```
