```python
from pymongo import MongoClient
from collections import defaultdict
from datetime import datetime
```

```python

client = MongoClient("mongodb://localhost:27017/")
db = client["Retail_Business"]

orders = db["Orders"]
stock_wh = db["stock_planning_warehouse"]
```

```python
raw_orders = list(orders.find())
len(raw_orders)
```

```python
stock_data = defaultdict(lambda: {
    "branch_name": None,
    "total_quantity": 0,
    "total_revenue": 0
})

for doc in raw_orders:
    branch = doc["branch"]["name"]

    row = stock_data[branch]
    row["branch_name"] = branch
    row["total_quantity"] += doc["product"]["quantity"]
    row["total_revenue"] += doc["sale"]["total_amount"]
```

```python
stock_wh.delete_many({})
stock_wh.insert_many(list(stock_data.values()))

stock_wh.count_documents({})
```
