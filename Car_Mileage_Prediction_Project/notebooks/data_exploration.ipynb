```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , r2_score
```

```python
df = pd.read_csv(r"D:\AI-and-Machine_Learning_Diploma\Car_Mileage_Prediction_Project\Dataset\auto-mpg.csv")
df

#
```

```python
plt.scatter(df['cylinders'],df['mpg'] , color = 'red' , alpha = 0.5)


plt.xlabel("Number of Cylinders")
plt.ylabel("Miles Per Gallon")

plt.title("Fuel Efficiency vs Cylinders")
plt.show()
```

```python
plt.scatter(df['displacement'],df['mpg'] , color = 'blue' , alpha = 0.5)


plt.xlabel("Engine Displacement")
plt.ylabel("Miles Per Gallon")

plt.title("Fuel Efficiency vs Engine Displacement")
plt.show()
```

```python

df["horsepower"] = pd.to_numeric(df["horsepower"], errors="coerce")
df = df.dropna(subset=["horsepower"])

bins = [0, 100, 150, 200, 300, 500]
labels = ["0-100", "101-150", "151-200", "201-300", "301+"]

df["hp_bin"] = pd.cut(df["horsepower"], bins=bins, labels=labels)

import seaborn as sns
sns.boxplot(x="hp_bin", y="mpg", data=df)
plt.xlabel("Horsepower Range")
plt.ylabel("Miles per Gallon")
plt.title("Fuel Efficiency by Horsepower Range")
plt.show()


# أضف jitter بسيط لقيم horsepower لتقليل التداخل
jitter = np.random.uniform(-2, 2, size=len(df))
hp_jitter = df["horsepower"] + jitter

plt.scatter(hp_jitter, df["mpg"], color="green", alpha=0.5)

plt.xlabel("Horsepower (with jitter)")
plt.ylabel("Miles per Gallon")
plt.title("Fuel Efficiency vs Horsepower (with jitter)")
plt.show()
```

```python
plt.show()
plt.scatter(df['weight'], df['mpg'], color='orange', alpha=0.5)

plt.xlabel("Vehicle Weight")
plt.ylabel("Miles per Gallon")

plt.title("Fuel Efficiency vs Vehicle Weight")
plt.show()

# Vehicle weight has a strong impact on miles per gallon.
```

```python
# acceleration
plt.show()
plt.scatter(df['acceleration'], df['mpg'], color='purple', alpha=0.5)

plt.xlabel("Acceleration")
plt.ylabel("Miles per Gallon")

plt.title("Fuel Efficiency vs Acceleration")
plt.show()

# Acceleration alone may not strongly affect MPG
```

```python
plt.show()
plt.scatter(df['model year'], df['mpg'], color='magenta', alpha=0.5)

plt.xlabel("Model Year")
plt.ylabel("Miles per Gallon")

plt.title("Fuel Efficiency vs Model Year")
plt.show()

# Model year usually has little or no effect on miles per gallon.
```

```python
plt.show()
plt.scatter(df['origin'], df['mpg'], color='teal', alpha=0.5)

plt.xlabel("Car Origin")
plt.ylabel("Miles per Gallon")
plt.title("Average Fuel Efficiency by Origin")
plt.show()

# Origin might have some effect, but it’s usually small on miles per gallon.
```

```python
plt.scatter(df['car name'], df['mpg'], color='magenta', alpha=0.5)

plt.xlabel("Car Name")
plt.ylabel("Average MPG")

plt.title("Fuel Efficiency vs Car Name ")
plt.show()

#Car name does not affect miles per gallon.
```
